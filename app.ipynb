```python
import os
import re
import numpy as np
import pandas as pd
import random

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
SEED = 42
np.random.seed(SEED)

random.seed(SEED)
```

```python
df = pd.read_csv("data/IMDB Dataset.csv")
print(f"Loaded {len(df):,} rows. Columns: {list(df.columns)}")
```

```python
print(df["sentiment"].value_counts())  # sanity check: should be ~25k/25k balanced
```

```python
def clean_text(text: str) -> str:
    """
    Cleans a single review string.

    The IMDB dataset is famous for containing raw HTML line breaks like "<br /><br />"
    left over from web scraping — these are pure noise for a classifier and MUST be
    stripped, or your model wastes capacity learning to ignore markup instead of
    learning sentiment signal.

    We deliberately do NOT do aggressive cleaning (e.g. removing punctuation or
    stopwords) here, because:
      - TF-IDF benefits mildly from punctuation removal, but we handle that inside
        the vectorizer itself (see TfidfVectorizer's token_pattern).
      - Transformers (DistilBERT) are trained on natural, punctuated text, so
        stripping punctuation would actually HURT the transformer's performance
        since it breaks the input distribution the pretrained model expects.
    """
    # Remove HTML tags (e.g. "<br /><br />")
    text = re.sub(r"<.*?>", " ", text)
    # Collapse repeated whitespace introduced by tag removal
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
df.head()
```

```python
df["review_clean"] = df["review"].apply(clean_text)
# df["label"] = (df["sentiment"].str.lower() == "positive").astype(int)
df["label"] = df["sentiment"].str.lower().map({"positive": 1, "negative": 0})
```

```python
df.head()
```

```python
X = df["review_clean"]
y = df["label"]
```

```python
# Stratified split preserves the pos/neg class balance in both sets — important
# even though this dataset is already 50/50, it's good practice in general.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y
)
```

```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=50_000,
    min_df=2,
    stop_words="english",
    sublinear_tf=True,  # log-scale term frequency, tends to help text classification
)
```

```python
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)  # transform only — never re-fit on test data
```

```python
clf = LogisticRegression(
    C=1.0,
    max_iter=1000,
    random_state=SEED,
    n_jobs=-1,
)
clf.fit(X_train_tfidf, y_train)
```

```python
y_pred = clf.predict(X_test_tfidf)

acc = accuracy_score(y_test, y_pred)
print(f"\nBaseline Test Accuracy: {acc:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["negative", "positive"]))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
def predict_baseline(text, clf, vectorizer):
    """
    Predicts the sentiment of a given text using the trained classifier and vectorizer.

    Args:
        text (str): The input text to classify.
        clf: The trained classifier.
        vectorizer: The fitted TF-IDF vectorizer.

    Returns:
        str: The predicted sentiment ("positive" or "negative").
    """
    # Clean the input text
    cleaned_text = clean_text(text)
    # Transform the cleaned text using the fitted vectorizer
    text_tfidf = vectorizer.transform([cleaned_text])
    # Predict the sentiment
    prediction = clf.predict(text_tfidf)[0]
    proba = clf.predict_proba(text_tfidf)[0]

    label = "positive" if prediction == 1 else "negative"

    class_index = list(clf.classes_).index(prediction)
    confidence = proba[class_index]
    # confidence = proba[1] if prediction == 1 else proba[0]

    return label, confidence
```
